2. A code cell is given that defines two lists of numbers, then attempts to define a Numpy array from the entries of the two lists using a mathematical operation. The code cell runs fine, but the output is different than what was intended. You have to add a comment, briefly explaining why the output of the code is doing something differently. Then run a corrected version of the code in another cell.

```python
#refer to class 5 

#Possible mistakes
# ------------------------------------------------------------------------------------------------------------
#1. 
import numpy as np

list1 = [1, 2, 3]
list2 = [4, 5, 6]

# Problematic line
result = list1 + list2  # ❌ This concatenates the lists, it does NOT add element-wise
print(result) # The '+' operator on Python lists concatenates them, it does not perform element-wise addition like Numpy arrays.

#$$$$$$$ Correction $$$$$$$$
# Convert lists to Numpy arrays for element-wise addition
arr1 = np.array(list1)
arr2 = np.array(list2)

result = arr1 + arr2  # ✅ Element-wise addition
print(result)
# ------------------------------------------------------------------------------------------------------------

#2. 
# Example arrays
A = np.array([1, 2, 3])
B = np.array([4, 5, 6])

# Elementwise addition
add_result = A + B          # [5 7 9]

# Elementwise subtraction
sub_result = A - B          # [-3 -3 -3]

# Elementwise multiplication
mult_result = A * B         # [4 10 18]

# Elementwise division
div_result = A / B          # [0.25 0.4 0.5]

# Elementwise exponentiation
exp_result = A ** 2         # [1 4 9]

# Matrix multiplication (dot product)
matmul_result = A @ B       # 1*4 + 2*5 + 3*6 = 32

# Display all results
print("Addition:", add_result)
print("Subtraction:", sub_result)
print("Multiplication:", mult_result)
print("Division:", div_result)
print("Exponentiation:", exp_result)
print("Matrix multiplication:", matmul_result)
```

4. Write a function that takes in a single number as an input argument, and returns a single string as the output. The value of the output string is one of two choices, which is determined by whether the input satisfies a logical condition. Then use the .apply() method from Pandas to evaluate this function for every row in a DataFrame. You will be told which column/variable should be input to the function.

```python

# Reffer to lesson 11 for help 

# Load CSV
df = pd.read_csv("your_file.csv")  # replace with your actual filename

# Function (as you defined)
def height_category(x):
    """
    Takes a single number x and returns:
    - "Tall" if x >= 70
    - "Short" otherwise
    """
    if x >= 70:
        return "Tall"
    else:
        return "Short"



# Apply the function to a column and save to a variable
ht_category = df['Height_inches'].apply(height_category)

# Print the result
print(ht_category.head())


```
